````python
# per ogni feature categorica, plottiamo la distribuzione delle feature continue 
# per vedere se ci sono pattern o differenze nelle distribuzioni

# per ogni feature categorica, creiamo un subplot con le feature numeriche selezionate come boxplot
selected_numerical_features = [f for f in numerical_features if f not in ['capital-gain', 'capital-loss']]
for cat_feature in X.columns:
    if is_categorical[cat_feature]:
        n_num = len(selected_numerical_features)
        fig, axes = plt.subplots(1, n_num, figsize=(8 * n_num, 4), sharey=False)
        for idx, num_feature in enumerate(selected_numerical_features):
            # Compute mean of the numerical feature for each category
            means = X.groupby(cat_feature)[num_feature].mean().sort_values()
            order = means.index.tolist()
            sns.boxplot(
                x=X[cat_feature],
                y=X[num_feature],
                ax=axes[idx],
                showfliers=False,  # ignore outliers
                order=order
            )
            axes[idx].set_title(f'{num_feature} by {cat_feature}')
            axes[idx].tick_params(axis='x', rotation=90)
        plt.tight_layout()
        plt.show()
````

<span style="color:red"><strong>TODO 9</strong></span>: di seguito vediamo la proporzione del target positivo tra elementi che appartengono alla stessa classe categorica. Cosa ci dice?

````python
import matplotlib.pyplot as plt
import seaborn as sns

cat_features = [f for f in X.columns if is_categorical[f]]
n_cat = len(cat_features)

n_cols = 4
n_rows = (n_cat + n_cols - 1) // n_cols  
fig, axes = plt.subplots(n_rows, n_cols, figsize=(3*n_cols, 10 * n_rows))
axes = axes.flatten()

for i, cat_feature in enumerate(cat_features):
    # calcola il rapporto True/False per ogni categoria
    ct = pd.crosstab(df[cat_feature], df[target_col])
    ratio = ct[True] / ct[False]
    ratio_df = ratio.to_frame(name="True/False Ratio").sort_values(by="True/False Ratio", ascending=False)
    sns.heatmap(
        ratio_df, cmap="coolwarm", annot=True, fmt=".2f", ax=axes[i], cbar=False
    )
    axes[i].set_title(f"Income Ratio by\n{cat_feature}")
    axes[i].set_xlabel("Ratio")
    axes[i].set_ylabel(cat_feature)
plt.tight_layout()
plt.show()
````

Questi plot mostrano il rapporto tra il numero di campioni positivi del target ("True") rispetto ai negativi ("False") per ciascuna categoria all’interno delle variabili categoriche.

Questo tipo di visualizzazione è utile per evidenziare l’associazione tra le singole categorie e il target, dal momento che non è possibile calcolare direttamente una correlazione numerica tra variabili categoriche e target binario.

````python
# Risposta 12:
def ols(X_train, y_train): 
    # Aggiungi una colonna di 1 per il termine di bias (in questo caso sarà il primo coefficiente)
    X_train = np.hstack((np.ones((X_train.shape[0], 1)), X_train))

    # Calcola i coefficienti usando la formula OLS: (X^T * X)^-1 * X^T * y
    beta = np.linalg.inv(X_train.T @ X_train) @ X_train.T @ y_train
    return beta

def predict(X_test, beta): 
    y_pred = beta[0] + X_test @ beta[1:]
    return y_pred


X = data[['temperatura', 'tempo']].to_numpy()
y = data['concentrazione'].to_numpy()

# per valutare correttamente, dobbiamo fare cross validation stratificata
mses, r2s = [], []
k_fold = 5 
for k in range(k_fold): 
    # dividiamo in train e test set stratificati
    X_train, X_test, y_train, y_test = random_train_test_split(X, y, proportion_train=0.8)

    # calcoliamo i coefficienti 
    beta = ols(X_train, y_train)

    # predizioni sul test set 
    y_pred = predict(X_test, beta)

    # calcoliamo l'errore quadratico medio e l'R2. Salviamoli nelle liste.
    mse = np.mean((y_test - y_pred) ** 2)
    r2 = 1 - np.sum((y_test - y_pred) ** 2) / np.sum((y_test - np.mean(y_test)) ** 2)

    mses.append(mse)
    r2s.append(r2)

# boxplot degli MSE e R2
plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.boxplot(mses)
plt.title("MSEs from Cross-Validation")
plt.xlabel("Mean Squared Error")
plt.subplot(1, 2, 2)
plt.boxplot(r2s)
plt.title("R2 Scores from Cross-Validation")
plt.xlabel("R2 Score")
plt.show()

# Risposta 12.1 (opzionale):
def ridge_regression(X_train, y_train, lambda_): 
    # Aggiungi una colonna di 1 per il termine di bias
    X_expanded = np.hstack((np.ones((X_train.shape[0], 1)), X_train))

    # Calcola i coefficienti usando la formula Ridge: (X^T * X + lambda_ * I)^-1 * X^T * y
    # Attenzione a non regolarizzare il termine di bias
    I = np.eye(X_expanded.shape[1])
    I[0, 0] = 0 
    beta = np.linalg.inv(X_expanded.T @ X_expanded + lambda_ * I) @ X_expanded.T @ y_train
    return beta

# NB: per la regressione ridge, dobbiamo normalizzare le feature! 
X = data[['temperatura', 'tempo']].to_numpy()
y = data['concentrazione'].to_numpy()

ols_mses, ols_r2s = [], []
ridge_mses, ridge_r2s = [], []

k_fold = 10
lambda_ = 100     # settiamolo a un valore arbitrario, ma si potrebbe usare nested cross-validation per trovare il migliore
for k in range(k_fold):
    # dividiamo in train e test set stratificati
    X_train, X_test, y_train, y_test = random_train_test_split(X, y, proportion_train=0.8)

    # normalizziamo le feature (non necessario su OLS, ma necessario su Ridge)
    # attenzione a usare le statistiche del training set per normalizzare il test set!
    X_train_mean = np.mean(X_train, axis=0)
    X_train_std = np.std(X_train, axis=0)
    X_train = (X_train - X_train_mean) / X_train_std
    X_test = (X_test - X_train_mean) / X_train_std

    # calcoliamo i coefficienti OLS
    beta_ols = ols(X_train, y_train)
    y_pred_ols = predict(X_test, beta_ols)

    # calcoliamo l'errore quadratico medio e l'R2 per OLS
    mse_ols = np.mean((y_test - y_pred_ols) ** 2)
    r2_ols = 1 - np.sum((y_test - y_pred_ols) ** 2) / np.sum((y_test - np.mean(y_test)) ** 2)

    ols_mses.append(mse_ols)
    ols_r2s.append(r2_ols)

    # calcoliamo i coefficienti Ridge
    beta_ridge = ridge_regression(X_train, y_train, lambda_)
    y_pred_ridge = predict(X_test, beta_ridge)

    # calcoliamo l'errore quadratico medio e l'R2 per Ridge
    mse_ridge = np.mean((y_test - y_pred_ridge) ** 2)
    r2_ridge = 1 - np.sum((y_test - y_pred_ridge) ** 2) / np.sum((y_test - np.mean(y_test)) ** 2)

    ridge_mses.append(mse_ridge)
    ridge_r2s.append(r2_ridge)

# boxplot degli MSE e R2 per OLS e Ridge
plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.boxplot([ols_mses, ridge_mses], tick_labels=['OLS', 'Ridge'])
plt.title("MSEs from Cross-Validation")
plt.xlabel("Mean Squared Error")
plt.subplot(1, 2, 2)
plt.boxplot([ols_r2s, ridge_r2s], tick_labels=['OLS', 'Ridge'])
plt.title("R2 Scores from Cross-Validation")
plt.xlabel("R2 Score")
plt.show()
````

````python
# Risposta 13.1:

def pairwise_distances(X, centroids): 
    """Returns the pairwise distance of each point in X with respect to each centroid.
    
    Args:
        X (np.ndarray): Data points of shape (n_samples, n_features).
        centroids (np.ndarray): Centroids of shape (n_centroids, n_features).
    """
    # distances = np.linalg.norm(X[:, np.newaxis] - self.centroids, axis=2) # more efficient but less intuitive
    pairwise_distances = np.zeros((X.shape[0], centroids.shape[0]))
    for i in range(X.shape[0]): 
        for j in range(centroids.shape[0]): 
            pairwise_distances[i, j] = np.linalg.norm(X[i] - centroids[j])
    return pairwise_distances

class KMeansClustering: 
    def __init__(self, n_clusters:int, max_iters:int=100, tol:float=1e-6): 
        """
        Args:
            n_clusters: int - the number of clusters.
            max_iters: int - maximum number of iterations.
            tol: float - tolerance for convergence.
            random_state: int - seed for reproducibility.
        """
        self.k = n_clusters
        self.max_iters = max_iters
        self.tol = tol
        self.centroids = None
        self.labels_ = None         

    def _assign_clusters(self, X): 
        distances = pairwise_distances(X, self.centroids)
        clusters = np.argmin(distances, axis=1)
        return clusters
    
    def _update_centroids(self, X, labels): 
        new_centroids = np.array([X[labels == i].mean(axis=0) for i in range(self.k)])
        return new_centroids

    def fit(self, X): 
        # inizializzazione casuale dei centroidi
        self.centroids = X[np.random.choice(X.shape[0], self.k, replace=False)]
        labels = self._assign_clusters(X)
        
        # iterare fino a convergenza o fino al numero massimo di iterazioni
        for _ in range(self.max_iters): 
            new_centroids = self._update_centroids(X, labels)
            if np.all(np.abs(new_centroids - self.centroids) < self.tol): 
                break
            self.centroids = new_centroids
            labels = self._assign_clusters(X)

        self.labels_ = labels
        return self
    

kmeans = KMeansClustering(n_clusters=3).fit(X)
fig = plt.figure(figsize=(8, 6))
plt.scatter(X[:, 0], X[:, 1], c=kmeans.labels_, cmap='viridis', marker='o', s=50)
plt.title("K-Means Clustering")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.legend()
plt.show()
````

## 3. Nuovo algoritmo: Versione più leggera di k-means

Nei passaggi precedenti, avete visto casi in cui è conveniente dividere il dataset in cluster e allenare classificatori separati per ciascun cluster. Per classificare un nuovo punto, è necessario associare il punto al cluster più vicino e poi utilizzare il classificatore corrispondente. A questo punto, il centroide del cluster associato dovrebbe essere aggiornato.

Tuttavia, l’aggiornamento dei cluster richiede tipicamente di memorizzare tutti i punti del dataset in memoria, il che aumenta significativamente l’uso della memoria rispetto all’allenamento di un singolo modello per l’intero dataset.

Per ovviare a questo, possiamo considerare versioni più leggere di k-means che non richiedono di conservare l’intero dataset in memoria.

### Batched k-means

Invece di memorizzare *tutti i punti* in memoria per aggiornare i centroidi, considera di mantenere solo un *sottoinsieme a dimensione fissa* dei dati.

In particolare:

- Definisci il budget di memoria come $M = m \times k$, dove: 




  - $m$ è un parametro regolabile (ad esempio, $m = 10$)  
  - $k$ è il numero di cluster  




- Implementa una variante di k-means in cui solo $M$ punti vengono salvati in memoria.

- Quando un nuovo punto viene assegnato a un cluster:  
  - Campiona nuovamente $M$ punti da mettere nel buffer, tra gli $M$ punti esistenti + quello nuovo.
  - Aggiorna il centroide utilizzando *solo i punti memorizzati nel buffer*.

````python
# Risposta 14.1:

class BatchedKMeans:
    def __init__(self, n_clusters: int, buffer_size: int, max_iters: int = 100, tol: float = 1e-6):
        """
        Inizializza l'oggetto BatchedKMeans.

        Args:
            n_clusters (int): Numero di cluster da creare.
            buffer_size (int): Dimensione fissa del buffer per ogni cluster. # CHANGE
            max_iters (int): Numero massimo di iterazioni.
            tol (float): Tolleranza per la convergenza.
            random_state (int): Seed per la riproducibilità.
        """
        self.k = n_clusters
        self.buffer_size = buffer_size                      # modifica
        self.max_iters = max_iters
        self.tol = tol
        self.centroids = None
        self.buffers = [[] for _ in range(self.k)]          # modifica. Ci salviamo un po' di punti per ogni cluster

    def _assign_clusters(self, X):
        """
        Assegna ogni punto al centroide più vicino.
        """
        distances = pairwise_distances(X, self.centroids)
        return np.argmin(distances, axis=1)

    def _update_centroids_from_buffers(self):
        """
        Aggiorna i centroidi facendo la media dei punti presenti nel buffer di ogni cluster.
        Bisogna chiamare questa funzione dopo aver aggiunto nuovi punti al buffer.
        """
        self.centroids = np.array([
            np.mean(self.buffers[i], axis=0) if len(self.buffers[i]) > 0 else self.centroids[i]
            for i in range(self.k)
        ])

    def fit(self, X):
        """
        Esegue il clustering K-means sui dati iniziali e inizializza i buffer.
        """
        # Inizializza i centroidi scegliendo k punti a caso
        self.centroids = X[np.random.choice(X.shape[0], self.k, replace=False)]
        labels = self._assign_clusters(X)

        for _ in range(self.max_iters):
            # standard: aggiorna i centroidi come la media dei punti assegnati a ciascun cluster
            new_centroids = np.array([X[labels == i].mean(axis=0) if np.any(labels == i) else self.centroids[i]
                                      for i in range(self.k)])

            if np.all(np.abs(new_centroids - self.centroids) < self.tol):
                break

            self.centroids = new_centroids
            labels = self._assign_clusters(X)

        # Inizializza i buffer di ogni cluster con subset casuale dei suoi punti
        for i in range(self.k):
            cluster_points = X[labels == i]
            if len(cluster_points) > 0:
                sampled = cluster_points[np.random.choice(len(cluster_points), min(self.buffer_size, len(cluster_points)), replace=False)]
                self.buffers[i] = list(sampled)
        return self

    def update_with_point(self, x_new):
        """
        Aggiorna il cluster più vicino aggiungendo un nuovo punto al buffer e ricalcolando il centroide.
        """
        x_new = np.asarray(x_new).reshape(1, -1)
        cluster_idx = self._assign_clusters(x_new)[0]

        # Aggiunge il nuovo punto al buffer del cluster
        self.buffers[cluster_idx].append(x_new[0])

        # Se il buffer eccede la capacità, riscampiona a dimensione fissa
        if len(self.buffers[cluster_idx]) > self.buffer_size:
            self.buffers[cluster_idx] = list(
                np.random.choice(self.buffers[cluster_idx], self.buffer_size, replace=False)
            )

        # Aggiorna solo il centroide del cluster modificato
        self._update_centroids_from_buffers()

    def predict(self, X): 
        # utility function opzionale
        return self._assign_clusters(X)
````

````python
# Risposta 15.1:
np.random.seed(1)

class OnlineKMeansClustering:
    def __init__(self, n_clusters: int, max_iters: int = 10, random_state: int = None):
        self.k = n_clusters
        self.max_iters = max_iters
        self.centroids = None        # per salvare i centroidi
        self.counts = None           # CHANGE: traccia il numero di update per ogni cluster

    def _assign_clusters(self, X): 
        """ Assegna ogni punto di X al centroide più vicino."""
        distances = pairwise_distances(X, self.centroids)
        return np.argmin(distances, axis=1)
    
    def _update_centroids(self, x, cluster_idx): 
        """
        Funzione che aggiorna il centroide di un cluster con un nuovo punto x, online. 
        Args:
            x (np.ndarray): Punto da aggiungere al cluster
            cluster_idx (int): Indice del cluster a cui appartiene il punto
        """
        count = self.counts[cluster_idx]                # numero di punti già assegnati a quel cluster 
        old_centroid = self.centroids[cluster_idx]
        
        # usiamo la regola della media incrementale: il nuovo centroide è una media pesata tra il vecchio
        # centoide, pesato per il numero di punti già assegnati al cluster, e il nuovo punto x, pesato per 1 
        # tutto diviso per il numero totale di punti nel cluster (count + 1)
        new_centroid = (count * old_centroid + x) / (count + 1)
        self.centroids[cluster_idx] = new_centroid
        self.counts[cluster_idx] += 1 

    def fit(self, X): 
        # inizializza random i centroidi 
        self.centroids = X[np.random.choice(X.shape[0], self.k, replace=False)]
        self.counts = np.zeros(self.k, dtype=int)           # inizializza il contatore dei punti per ogni centroide/cluster
        
        for _ in range(self.max_iters): 
            # assegna i punti ai cluster
            labels = self._assign_clusters(X)

            # aggiorna i centroidi con i punti assegnati
            for i in range(X.shape[0]): 
                self._update_centroids(X[i], labels[i])
````
